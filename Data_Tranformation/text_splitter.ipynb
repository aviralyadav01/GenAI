```python
#1st load the document which is to be divided into chunks 
# ye document load hoga as a default type of DOCUMENTS 
from langchain_docling.loader import DoclingLoader
FILE_PATH= "data.txt"
loader = DoclingLoader(file_path = FILE_PATH)
doc = loader.load()
doc
```

```python
#text splitter 
#using RECURSIVE CHARACTER TEXT SPLITTER 
from langchain_text_splitters import RecursiveCharacterTextSplitter

with open("data.txt") as f:
    data = f.read()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 50 ,
    chunk_overlap = 10, )
text = text_splitter.create_documents([data])
print(text[0])
print(text[1])
```

```python
# TEXT SPLITTING ON THE BASIS ON OUR SEPERATOR 

from langchain_text_splitters import CharacterTextSplitter
text1 = CharacterTextSplitter(separator = "\n\n" , chunk_size = 50 , chunk_overlap = 10 ,)
final_text = text1.split_documents(doc)
print(final_text[0])
```
